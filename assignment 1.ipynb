```python
#dictionary
sdictionary={
    101:"drishti",
    102:"sakshi",
    103:"simran",
    104:"chetana",
    105:"vedanshi"
}
print("student dictionary")
print(sdictionary)
```

```python
#add
sdictionary[106]="aarya"
print("after adding a student")
print(sdictionary)
```

```python
#update
sdictionary[101]="drishti israni"
print("after updating in list")
print (sdictionary)
```

```python
#delete
del sdictionary[104]
print("after deleting a student")
print(sdictionary)
```
