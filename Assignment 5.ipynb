```python
#check that the strings contains certain set of characters
import re

def check_strings(s):
    pattern = "^[A-Za-z0-9]+$" #^=start,$=end,+=one or more
    return bool(re.match(pattern,s))

#Example usage
s1="Hello123"
s2="Hello_123"  #contains underscore

print(check_strings(s1))
print(check_strings(s2))
```
