```python
#TO CHECK WHETHER OR NOT THE TRIANGLE IS A RIGHT-ANGLED TRIANGLE USING FUNCTION
def right_triangle(a,b,c):
    if a*a+b*b==c*c:
        print("Right angled triangle")
    else:
        print("Not a right angled triangle")
right_triangle(3,4,5)
```
